Questão 1: Analisando temperaturas.

```python
temperaturas = []

for i in range(7):
    temperatura = float(input("Temperatura: "))
    temperaturas.append(temperatura)

soma = 0

for t in temperaturas:
    soma += t

media = soma / len(temperaturas)

acima = 0
abaixo = 0
iguais = 0

for t in temperaturas:
    if t > media:
        acima += 1
    elif t < media:
        abaixo += 1
    else:
        iguais += 1

print("Temperaturas:", temperaturas)
print("Média:", media)
print("Acima:", acima)
print("Abaixo:", abaixo)
print("Iguais:", iguais)
```

Questão 2: Ponto e distância da origem.

```python

x = int(input("x: "))
y = int(input("y: "))

ponto = (x, y)

if ponto[0] == 0 and ponto[1] == 0:
    local = "Origem"
elif ponto[0] == 0:
    local = "Eixo Y"
elif ponto[1] == 0:
    local = "Eixo X"
elif ponto[0] > 0 and ponto[1] > 0:
    local = "1º quadrante"
elif ponto[0] < 0 and ponto[1] > 0:
    local = "2º quadrante"
elif ponto[0] < 0 and ponto[1] < 0:
    local = "3º quadrante"
else:
    local = "4º quadrante"

dist2 = ponto[0] ** 2 + ponto[1] ** 2

print("Ponto:", ponto)
print("Local:", local)
print("Distância²:", dist2)
```

Questão 4: Maior sequência de crescimento.

```python
numeros = []
for i in range(10):
    numero = int(input("Número: "))
    numeros.append(numero)
atual = 1
maior = 1
for i in range(1, len(numeros)):
    if numeros[i] > numeros[i - 1]:
        atual += 1
    else:
        atual = 1
    if atual > maior:
        maior = atual
print("Lista:", numeros)
print("Maior sequência crescente:", maior)
```

Questão 5: Ponto no plano.

```python
pontos = []
for i in range(6):
    x = int(input("x: "))
    y = int(input("y: "))
    ponto = (x, y)
    pontos.append(ponto)
q1 = 0
q2 = 0
q3 = 0
q4 = 0
maior_dist2 = -1
maior_ponto = None
for ponto in pontos:
    x = ponto[0]
    y = ponto[1]
    if x > 0 and y > 0:
        q1 += 1
    elif x < 0 and y > 0:
        q2 += 1
    elif x < 0 and y < 0:
        q3 += 1
    elif x > 0 and y < 0:
        q4 += 1
    dist2 = x ** 2 + y ** 2
    if dist2 > maior_dist2:
        maior_dist2 = dist2
        maior_ponto = ponto
print("Pontos:", pontos)
print("1º quadrante:", q1)
print("2º quadrante:", q2)
print("3º quadrante:", q3)
print("4º quadrante:", q4)
print("Ponto com maior x² + y²:", maior_ponto)
print("Maior valor de x² + y²:", maior_dist2)
```

Questão 7: Boletim de uma turma

```python
alunos = []
for i in range(6):
    nome = input("Nome: ")
    n1 = float(input("Nota 1: "))
    n2 = float(input("Nota 2: "))
    n3 = float(input("Nota 3: "))
    media = (n1 + n2 + n3) / 3
    aluno = {
        "nome": nome,
        "media": media
    }
alunos.append(aluno)
soma = 0
for aluno in alunos:
    soma += aluno["media"]
media_turma = soma / 6
quantidade = 0
for aluno in alunos:
    if aluno["media"] >= media_turma:
        quantidade += 1
maior = alunos[0]["media"]
for aluno in alunos:
    if aluno["media"] > maior:
        maior = aluno["media"]
print("\nAlunos e médias:")
for aluno in alunos:
    print(aluno["nome"], aluno["media"])
print("Média da turma:", media_turma)
print("Quantidade acima ou igual à média:", quantidade)
print("Aluno com maior média:")
for aluno in alunos:
    if aluno["media"] == maior:
        print(aluno["nome"])
```

Questão 8: Controle de estoque com movimentações

```python
produtos = []

for i in range(5):
    codigo = int(input("Digite o codigo: "))
    nome = input("Digite o nome: ")
    preco = float(input("Digite o preco: "))
    quantidade = int(input("Digite a quantidade: "))

    produtos.append({
        "codigo": codigo,
        "nome": nome,
        "preco": preco,
        "quantidade": quantidade
    })

for i in range(5):
    codigo = int(input("Digite o codigo do produto: "))
    tipo = input("Digite E para entrada ou S para saida: ")
    qtd = int(input("Digite a quantidade: "))

    for produto in produtos:
        if produto["codigo"] == codigo:
            if qtd > 0:
                if tipo == "E":
                    produto["quantidade"] = produto["quantidade"] + qtd

                elif tipo == "S":
                    if qtd <= produto["quantidade"]:
                        produto["quantidade"] = produto["quantidade"] - qtd

print("\nEstoque atualizado:")

total = 0

for produto in produtos:
    print("Codigo:", produto["codigo"])
    print("Nome:", produto["nome"])
    print("Preco:", produto["preco"])
    print("Quantidade:", produto["quantidade"])

    total = total + produto["preco"] * produto["quantidade"]

print("Valor total do estoque:", total)
```

Questão 9: Ranking de jogadores

```python
jogadores = []

for i in range(5):
    nome = input("Nome: ")
    notas = []

    for j in range(3):
        notas.append(int(input("Pontuação: ")))

    jogadores.append({"nome": nome, "notas": notas})

for jogador in jogadores:
    n = jogador["notas"]

    maior = n[0]
    menor = n[0]

    for x in n:
        if x > maior:
            maior = x
        if x < menor:
            menor = x

    jogador["total"] = n[0] + n[1] + n[2]
    jogador["media"] = jogador["total"] / 3
    jogador["dif"] = maior - menor

maior = jogadores[0]["total"]
regular = jogadores[0]["dif"]

for jogador in jogadores:
    if jogador["total"] > maior:
        maior = jogador["total"]
    if jogador["dif"] < regular:
        regular = jogador["dif"]

for jogador in jogadores:
    print(jogador["nome"], "Total:", jogador["total"], "Média:", jogador["media"])

print("Maior total:")
for jogador in jogadores:
    if jogador["total"] == maior:
        print(jogador["nome"])

print("Mais regulares:")
for jogador in jogadores:
    if jogador["dif"] == regular:
        print(jogador["nome"])
```

Questão 10: Agenda de contatos sem duplicidade

```python
contatos = []

while True:
    nome = input("Digite o nome do contato (ou FIM para parar): ")
    if nome == "FIM":
        break
    existe = False
    for contato in contatos:
        if contato["nome"] == nome:
            existe = True
    if existe:
        print("Esse nome já está cadastrado!")
    else:
        telefone = input("Digite o telefone: ")
        contato = {
            "nome": nome,
            "telefone": telefone
        }
        contatos.append(contato)
print("\n Consulta")
while True:
    nome = input("Digite um nome para consultar (ou FIM para sair): ")
    if nome == "FIM":
        break
    encontrado = False
    for contato in contatos:
        if contato["nome"] == nome:
            print("Telefone:", contato["telefone"])
            encontrado = True
    if not encontrado:
        print("Contato não encontrado.")
```

Questão 11: Pesquisa de votos

```python
votos = {
    1: 0,
    2: 0,
    3: 0,
    4: 0
}
while True:
    codigo = int(input("Digite o código do voto (1 a 4) ou 0 para encerrar: "))
    if codigo == 0:
        break
    if codigo >= 1 and codigo <= 4:
        votos[codigo] += 1
    else:
        print("Código inválido!")
total = votos[1] + votos[2] + votos[3] + votos[4]
print("\n Resultado")
for opcao in votos:
    percentual = (votos[opcao] / total) * 100 if total > 0 else 0
    print("Opção", opcao, ":", votos[opcao], "votos -", percentual, "%")
maior = votos[1]
for opcao in votos:
    if votos[opcao] > maior:
        maior = votos[opcao]
print("\nOpção(ões) vencedora(s):")
for opcao in votos:
    if votos[opcao] == maior:
        print("Opção", opcao)
```

Questão 12: Pesquisa de votos

```python
estoque = {
    1: 10,
    2: 20,
    3: 15,
    4: 8
}
historico = []
for i in range(8):
    codigo = int(input("Código do produto: "))
    tipo = input("Tipo (entrada ou saida): ")
    quantidade = int(input("Quantidade: "))
    movimentacao = (codigo, tipo, quantidade)
    if codigo not in estoque:
        print("Operação recusada!")
        historico.append((movimentacao, "Recusada"))
    elif tipo == "entrada":
        estoque[codigo] = estoque[codigo] + quantidade
        historico.append((movimentacao, "Aceita"))
    elif tipo == "saida":
        if estoque[codigo] >= quantidade:
            estoque[codigo] = estoque[codigo] - quantidade
            historico.append((movimentacao, "Aceita"))
        else:
            print("Operação recusada! Estoque insuficiente.")
            historico.append((movimentacao, "Recusada"))
    else:
        print("Tipo inválido!")
        historico.append((movimentacao, "Recusada"))
print("\nEstoque final")
for codigo in estoque:
    print(codigo, "-", estoque[codigo])
print("\nHistórico")
for movimentacao in historico:
    print(movimentacao)
```

Questão 12: Histórico de movimentações

```python
estoque = {
    101: 10,
    102: 5,
    103: 20
}
historico = []
movimentacoes = [
    (101, "entrada", 5),
    (102, "saida", 3),
    (103, "saida", 10),
    (101, "saida", 20),
    (102, "entrada", 8),
    (103, "entrada", 5),
    (101, "saida", 4),
    (102, "saida", 20)
]
for movimento in movimentacoes:
    codigo, tipo, quantidade = movimento
    if codigo not in estoque:
        historico.append((codigo, tipo, quantidade, "recusada"))
        continue
    if tipo == "entrada":
        estoque[codigo] += quantidade
        historico.append((codigo, tipo, quantidade, "realizada"))
    elif tipo == "saida":
        if estoque[codigo] >= quantidade:
            estoque[codigo] -= quantidade
            historico.append((codigo, tipo, quantidade, "realizada"))
        else:
            historico.append((codigo, tipo, quantidade, "recusada"))
print("Estoque final:")
for codigo, quantidade in estoque.items():
    print(codigo, ":", quantidade)
print("\nHistórico:")
for movimento in historico:
    print(movimento)
```

Questão 13: Análise de vendas por vendedor

```python
vendedores = []
for i in range(5):
    nome = input("Nome do vendedor: ")
    vendas = []
    for j in range(4):
        valor = float(input("Valor da venda: "))
        vendas.append(valor)
    vendedor = {
        "nome": nome,
        "vendas": vendas
    }
    vendedores.append(vendedor)
total_geral = 0
for vendedor in vendedores:
    total = 0
    for venda in vendedor["vendas"]:
        total = total + venda
    vendedor["total"] = total
    total_geral = total_geral + total
media = total_geral / 5
maior_total = vendedores[0]["total"]
for vendedor in vendedores:
    if vendedor["total"] > maior_total:
        maior_total = vendedor["total"]
print("\nRelatório")
for vendedor in vendedores:
    print(vendedor["nome"], "- Total vendido: R$", vendedor["total"])
print("Total geral: R$", total_geral)
print("Média da equipe: R$", media)
print("\nVendedor(es) com maior total:")
for vendedor in vendedores:
    if vendedor["total"] == maior_total:
        print(vendedor["nome"])
acima_media = 0
for vendedor in vendedores:
    if vendedor["total"] > media:
        acima_media = acima_media + 1
print("Quantidade acima da média:", acima_media)
```

Questão 14: Catálogo e pedidos

```python
produtos = []
for i in range(5):
    codigo = int(input("Código do produto: "))
    nome = input("Nome do produto: ")
    preco = float(input("Preço do produto: "))
    estoque = int(input("Quantidade em estoque: "))
    produto = {
        "codigo": codigo,
        "nome": nome,
        "preco": preco,
        "estoque": estoque
    }
    produtos.append(produto)
pedidos = []
for i in range(6):
    codigo = int(input("Código do pedido: "))
    quantidade = int(input("Quantidade: "))
    pedido = (codigo, quantidade)
    pedidos.append(pedido)
total_pedidos = 0
print("\nResultado dos pedidos")
for pedido in pedidos:
    codigo = pedido[0]
    quantidade = pedido[1]
    encontrado = False
    for produto in produtos:
        if produto["codigo"] == codigo:
            encontrado = True
            if produto["estoque"] >= quantidade:
                produto["estoque"] = produto["estoque"] - quantidade
                valor = produto["preco"] * quantidade
                total_pedidos = total_pedidos + valor
                print("Pedido atendido:", codigo, "-", quantidade)
            else:
                print("Pedido recusado:", codigo, "- estoque insuficiente")
    if encontrado == False:
        print("Pedido recusado:", codigo, "- código não encontrado")
print("\nValor total dos pedidos atendidos: R$", total_pedidos)
```
